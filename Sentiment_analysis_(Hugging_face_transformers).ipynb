# **Model Selection:**
**Selected Model:** cardiffnlp/twitter-roberta-base-sentiment-latest

I chose this model because this model is suitable for Twitter/social media texts sentiment analysis, this model contain 3 sentiment labels (Positive/Negative/Neutral), and important of all we didn't need to train model from scratch it is pre-trained.

# **Sample Dataset:**
I created this dataset to test the model with different text/comments. It includes all type of sentiments (Positive/Negative,Neutral). # **Initial Model Testing:**
First we will test the model with 3 cases.


```python
!pip install -q transformers torch
```

```python
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest"
)
```

```python
result = sentiment_pipeline("You are doing great work.")
result
```

```python
result = sentiment_pipeline("Well I don't trust this.")
result
```

```python
result = sentiment_pipeline("Does this organization work on the charity given by people?")
result
```

# **Initial test results:**
I tested first three datasets on model, positive/negative/neutral were included to check that model is not only detecting positive and negative it can also detect neutral and other type of comments and good thing is that prediction matched the same expected sentiments, it also produced score for each prediction.

# **Test Cases:**

```python
test_comments = [
    "You are doing great work.",
    "Well I don't trust this.",
    "Does this organization work on the charity given by people?",
    "Yeah I can see how much work is done.",
    "I'm happy this type of people exist who work for other people.",
    "Thank you for supporting our community and helping those in need.",
    "I am disappointed with how this organization handled the donations.",
    "Where can I find information about your next campaign?",
    "Another campaign, but nothing ever seems to change.",
    "The volunteers have done an amazing job helping local families."
]
```

```python
results = sentiment_pipeline(test_comments)
results
```
